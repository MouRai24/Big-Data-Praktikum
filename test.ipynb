```python
import pandas as pd

# CSV in DataFrame laden
df = pd.read_csv('linkage_problems/Music/train_pairs_fv.csv')
```

```python
pd.set_option('display.max_columns', None)     # keine Begrenzung der Spaltenzahl :contentReference[oaicite:0]{index=0}
display(df)
```

```python
# 2. Liste der Similarity-Spalten definieren
sim_cols = [
    "language_lev", "language_jaccard", "language_relaxed_jaccard",
    "language_overlap", "language_containment", "language_token_jaccard",
    "number_sim", "number_num_equal",
    "album_lev", "album_jaccard", "album_relaxed_jaccard",
    "album_overlap", "album_containment", "album_token_jaccard",
    "year_sim", "year_num_equal",
    "length_sim", "length_num_equal",
    "title_lev", "title_jaccard", "title_relaxed_jaccard",
    "title_overlap", "title_containment", "title_token_jaccard",
    "artist_lev", "artist_jaccard", "artist_relaxed_jaccard",
    "artist_overlap", "artist_containment", "artist_token_jaccard",
]

```

```python
# 3. Schwellenwerte festlegen
delta_H = 0.8
delta_L = 0.1

# 4. Labeling-Funktion definieren
def lambda1(row):
    # Nur gültige Werte (>= 0) mitteln; -1 = kein Vergleich
    vals = [v for v in row[sim_cols] if v >= 0]
    if len(vals) == 0:
        return -1
    avg = sum(vals) / len(vals)
    if avg >= delta_H:
        return 1
    elif avg <= delta_L:
        return 0
    else:
        return -1
```

```python
# 5. Funktion auf jedes Paar anwenden
df["lambda1_label"] = df.apply(lambda1, axis=1)

```

```python
# 6. Ergebnis anschauen
print(df[["pair_id", "lambda1_label"]].head(100))
```
